```python
from langchain_community.document_loaders import PyPDFLoader
from pathlib import Path
```

```python
all_files = list(Path("data").rglob("*.pdf"))
print(f"Found {len(all_files)} PDF files in the data directory.")
```

```python
document_objects = []
for pdf in all_files:
    print(f"Loading {pdf}...")
    loader = PyPDFLoader(pdf)
    document_objects.extend(loader.load())
```

```python
print(len(all_files)) 
print(len(document_objects)) 
```

```python
for document in document_objects:
    print(f"Document metadata: {document.metadata}")
```

```python
from langchain_text_splitters import (RecursiveCharacterTextSplitter,CharacterTextSplitter)
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_text(document_objects[0].page_content)
print(chunks \n)
print(f"Number of chunks: {len(chunks)}")
```
